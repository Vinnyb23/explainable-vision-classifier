# Phase 2 Exploration -- Explainable Vision Classifier

Week 7 scratchpad: look at sample images from both classes, sanity-check the dataset (real or synthetic fallback), and eyeball a few Grad-CAM heatmaps before those pieces get promoted into `src/`.

Run `python -m src.data_prep` first if `../data/casting_synthetic/` does not exist yet (or drop the real Kaggle dataset into `../data/casting/` -- see README "Get the real dataset").

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import matplotlib.pyplot as plt
from PIL import Image

from src.data_prep import get_dataset_dir, CLASS_NAMES

dataset_dir, using_real = get_dataset_dir()
print(f"Dataset dir: {dataset_dir}")
print(f"Using real Kaggle data: {using_real}")
```

```python
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
for row, cls in enumerate(CLASS_NAMES):
    cls_dir = os.path.join(dataset_dir, cls)
    files = sorted(os.listdir(cls_dir))[:4]
    for col, fname in enumerate(files):
        img = Image.open(os.path.join(cls_dir, fname))
        axes[row, col].imshow(img)
        axes[row, col].set_title(cls)
        axes[row, col].axis('off')
plt.tight_layout()
plt.show()
```

```python
from src.data_prep import build_dataloaders

train_dl, val_dl, test_dl, classes, using_real = build_dataloaders()
print(f"Classes: {classes}")
print(f"Train / val / test sizes: {len(train_dl.dataset)} / {len(val_dl.dataset)} / {len(test_dl.dataset)}")
```

## Grad-CAM sanity check

Run this after training at least one model (`python -m src.train --backbone resnet50`).

```python
from src.gradcam import load_model_for_explain, explain_image

backbone = "resnet50"
weights_path = os.path.join("..", "models", f"{backbone}.pt")

if os.path.exists(weights_path):
    model = load_model_for_explain(backbone, weights_path)
    sample_dir = os.path.join(dataset_dir, "def_front")
    sample_path = os.path.join(sample_dir, sorted(os.listdir(sample_dir))[0])

    pred, conf, overlay = explain_image(model, backbone, Image.open(sample_path))
    print(f"Prediction: {pred} (confidence={conf:.3f})")

    fig, ax = plt.subplots(1, 2, figsize=(8, 4))
    ax[0].imshow(Image.open(sample_path))
    ax[0].set_title("Input")
    ax[0].axis('off')
    ax[1].imshow(overlay)
    ax[1].set_title("Grad-CAM")
    ax[1].axis('off')
    plt.show()
else:
    print(f"No weights at {weights_path} yet -- train a model first.")
```
